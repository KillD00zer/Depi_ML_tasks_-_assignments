### Ecommerce shoping cart with OOP

```python
class ShoppingCart:
    def __init__ (self, items=[] , prices=[] ):
        self.items = items
        self.prices = prices

    def add_item(self, item, price):
        self.items.append(item)
        self.prices.append(price)
        print(f"Added {item} to cart.")
    
    def checkout(self):
        total = sum(self.prices)
        print(f"Total amount: {total}")
        print("Thank you for shopping with us!")
```

```python
tata= ShoppingCart()
tata.add_item("apple", 2)
tata.add_item("banana", 3)
tata.add_item("chips", 5)
tata.checkout()
```

### Social media profile OOP

```python
class Profile:
    def __init__ (self, name, gender, country):
        self.name = name
        self.gender = gender
        self.country = country
        self.hash_code= hash(self.name)

    def message(self , user_2 , message):
        print(f'{self.name}:[{self.hash_code}] sent this message:{message} to {user_2.name}:[{hash(user_2)}] ')
```

### Moive ticket booking

```python
class ticket_shop:
    def __init__(self, capacity, movies=[], tickets=[]):
        self.capacity = capacity
        self.movies = movies
        self.tickets = tickets

    def add_movie(self, movie):
        self.tickets = [i for i in range(self.capacity)]
        self.movies.append(movie)
        print(f'{movie} was added')
    
    def buy_ticket(self,movie,tickts_num):
        for i in range (tickts_num):
           self.tickets.pop()
        print(f'successfully bought {tickts_num} tickets for {movie}, remaining tickets: {len(self.tickets)}')
```

```python
Golden_cima = ticket_shop(200)
Golden_cima.add_movie("Avatar")
Golden_cima.add_movie("GodFather")

Golden_cima.buy_ticket("Avatar" , 1)
Golden_cima.buy_ticket("Avatar" , 3)

Golden_cima.buy_ticket('GodFather', 150)
```
